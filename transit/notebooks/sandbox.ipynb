```python
from google.transit import gtfs_realtime_pb2
from datetime import datetime, timezone

feed = gtfs_realtime_pb2.FeedMessage()

with open("../data/TripUpdates (3)", "rb") as f:
    feed.ParseFromString(f.read())

print(feed.header.gtfs_realtime_version)
print(datetime.fromtimestamp(feed.header.timestamp, timezone.utc))
print(len(feed.entity))

for entity in feed.entity[:5]:
    tu = entity.trip_update
    print("trip_id:", tu.trip.trip_id)
    print("route_id:", tu.trip.route_id)
    print("start_date:", tu.trip.start_date)
    print("vehicle:", tu.vehicle.id)

    for stu in tu.stop_time_update:
        print({
            "stop_id": stu.stop_id,
            "stop_sequence": stu.stop_sequence,
            "arrival_time": datetime.fromtimestamp(stu.arrival.time, timezone.utc)
                if stu.HasField("arrival") and stu.arrival.HasField("time") else None,
            "arrival_delay": stu.arrival.delay
                if stu.HasField("arrival") and stu.arrival.HasField("delay") else None,
        })
```

```python
feed = gtfs_realtime_pb2.FeedMessage()

with open("../data/VehiclePositions", "rb") as f:
    feed.ParseFromString(f.read())

feed.entity[90:100]
```

```python
from collections import Counter

# Collect trip_ids from TripUpdate entities
trip_ids = [
    entity.trip_update.trip.trip_id
    for entity in feed.entity
    if entity.HasField("trip_update") and entity.trip_update.trip.trip_id
]

total_trip_ids = len(trip_ids)
unique_trip_ids = len(set(trip_ids))

print(f"Total trip_id values: {total_trip_ids}")
print(f"Unique trip_id values: {unique_trip_ids}")
print(f"All trip_id values unique? {total_trip_ids == unique_trip_ids}")

if total_trip_ids != unique_trip_ids:
    counts = Counter(trip_ids)
    duplicate_trip_ids = {trip_id: count for trip_id, count in counts.items() if count > 1}
    print("\nDuplicate trip_ids and counts:")
    for trip_id, count in sorted(duplicate_trip_ids.items()):
        print(f"  {trip_id}: {count}")
else:
    print("\nNo duplicate trip_ids found.")
```
